```python
from crewai import Agent, Task, Crew, Process, LLM
from dotenv import load_dotenv
from email_reader_tool import EmailReaderTool

load_dotenv()

llm = LLM(model="groq/llama-3.3-70b-versatile")

reader_agent = Agent(
    role="Email Reader",
    goal="Read unread emails",
    backstory="You extract email details.",
    tools=[EmailReaderTool()],
    llm=llm,
    verbose=True
)

intent_agent = Agent(
    role="Email Intent Analyzer",
    goal="Understand what the email is about",
    backstory="You classify emails by intent.",
    llm=llm,
    verbose=True
)

task1 = Task(
    description="Read unread emails and extract sender, subject, and content.",
    expected_output="List of emails with details.",
    agent=reader_agent
)

task2 = Task(
    description="""
For each email, classify the intent as one of:
- Question
- Meeting
- Request
- General

Give a short reason.
""",
    expected_output="Emails with classified intent.",
    agent=intent_agent,
    context=[task1]
)

crew = Crew(
    agents=[reader_agent, intent_agent],
    tasks=[task1, task2],
    process=Process.sequential,
    verbose=True,
    tracing=True
)

result = crew.kickoff()
print("\n=== EMAIL INTENT ANALYSIS ===\n")
print(result)
```
